````python
# Importe a biblioteca OpenAI
from openai import OpenAI
# Usado para armazenar com segurança sua chave de API
from google.colab import userdata

# Configure a chave de API da OpenAI
OPENAI_API_KEY = userdata.get('OPENAI_API_KEY')

# Inicialize o cliente OpenAI
openai_client = OpenAI(api_key=OPENAI_API_KEY)
````

````python
# Defina a função para gerar a mensagem usando OpenAI
def generate_openai_message(user):
    prompt = f"Crie uma mensagem para {user['nome']} agradecendo a avaliação do evento Festival da Costela de acordo com oq ela avaliou. Máximo de 150 caracteres."

    response = openai_client.chat.completions.create(
        model="gpt-3.5-turbo",  # Você pode escolher outro modelo OpenAI aqui, se preferir
        messages=[
            {"role": "system", "content": "Você é um assistente prestativo."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=60 # Ajuste conforme necessário para 150 caracteres
    )
    return response.choices[0].message.content.strip()

# Adapte o loop para usar a nova função
for user in df.to_dict('records'):
    try:
        message = generate_openai_message(user)
        print(f"Mensagem para {user['nome']}: {message}")
    except Exception as e:
        print(f"Erro ao gerar mensagem para {user['nome']}: {e}")
````

````python
# CRIANDO PERFIL DOS AVALIADORES

perfis = []

for _, usuario in df.iterrows():

    perfil = {
        'id': int(usuario['id']),
        'nome': usuario['nome'],
        'cidade': usuario['cidade'],
        'bairro': usuario['bairro'],
        'faixa_etaria': usuario['faixa_etaria'],
        'participou_outras_edicoes': usuario['participou_outras_edicoes'],
        'como_soube': usuario['como_soube'],
        'motivo_participacao': usuario['motivo_participacao'],
        'gasto': usuario['gasto'],
        'pretende_retornar': usuario['pretende_retornar'],
        'recomendacao': int(usuario['recomendacao'])
    }
    perfis.append(perfil)
````

## CRIANDO PERFIL COMPORTAMENTAL

````python
def definir_perfil(usuario):

    participou = usuario['participou_outras_edicoes']
    retorno = usuario['pretende_retornar']
    recomendacao = usuario['recomendacao']

    if participou == 'Sim' and retorno == 'Sim':
        return 'Fiel'

    elif participou == 'Não' and retorno == 'Sim':
        return 'Novo visitante'

    elif participou == 'Sim' and retorno == 'Talvez':
        return 'Potencial recorrente'

    elif retorno == 'Não':
        return 'Baixa intenção de retorno'

    elif recomendacao >= 8:
        return 'Promotor'

    else:
        return 'Indefinido'

def definir_fidelidade(usuario):

    participou = usuario['participou_outras_edicoes']
    retorno = usuario['pretende_retornar']

    if participou == 'Sim' and retorno == 'Sim':
        return 'Alta'

    elif participou == 'Sim' and retorno == 'Talvez':
        return 'Média'

    elif participou == 'Não' and retorno == 'Sim':
        return 'Baixa'

    else:
        return 'Muito baixa'

def definir_potencial_consumo(usuario):

    gasto = usuario['gasto']

    if gasto == 'Acima de R$200':
        return 'Alto'

    elif 'R$100' in gasto or '100' in gasto:
        return 'Médio'

    elif 'R$50' in gasto or '50' in gasto:
        return 'Baixo'

    else:
        return 'Não identificado'

def definir_engajamento(usuario):

    recomendacao = usuario['recomendacao']

    if recomendacao >= 9:
        return 'Alto'

    elif recomendacao >= 7:
        return 'Médio'

    else:
        return 'Baixo'
````

````python
for perfil in perfis:

    perfil['fidelidade'] = definir_fidelidade(perfil)
    perfil['potencial_consumo'] = definir_potencial_consumo(perfil)
    perfil['nivel_engajamento'] = definir_engajamento(perfil)
````
